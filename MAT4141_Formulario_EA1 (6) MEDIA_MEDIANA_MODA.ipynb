### Tabla de frecuencia para variable **cualitativa** de un DataFrame de nombre **df**

```python
#Cálculo de frecuencia absoluta. La palabra frecuencia representa la frecuencia absoluta. (fi)
frecuencia = df.groupby('Nombre de la variable').size()


    

#Crear el DF con frecuencias absolutas y relativas
tabla = pd.DataFrame({
    'fi': frecuencia,
    'hi (%)': round((frecuencia/muestra) *100, 1) # redondear según contexto y valores de los datos.
})
tabla.loc['Total'] = [tabla['fi'].sum(), tabla['hi (%)'].sum()]
tabla










#cuando_sea_solicitado
#La tabla es ordenada de mayor a menor, ascending= False,
#por los valores de frecuencia absoluta y se muestran los primeros 10 valores
tabla.sort_values('fi', ascending= False).head(10)
```

## TABLA FRECUENCIA COMPLETA

```python
#Frecuencia absoluta:
frecuencia = df.groupby('Asignaturas reprobadas').size()

#Tabla con 4 frecuencias completa
tabla= pd.DataFrame({
    'fi': frecuencia,
    'hi (%)': frecuencia/muestra * 100,
    'FI': frecuencia.cumsum(),
    'HI (%)': (frecuencia/muestra * 100).cumsum()
})
tabla
```

# TABLA DE FRECUENCIA CON INTERVALOS

```python
#Frecuencia absoluta:
frecuencia = df.groupby('XXXXX').size()

# Crear los 8 intervalos
df['Intervalos_XXX'] = pd.cut(df['XXXXX'], bins = 8, include_lowest= True, precision = 2)

# Frecuencia absoluta por intervalos
frecuencia = df.groupby(
    'Intervalos_XXX',
    observed=True
).size()

# Tamaño de la muestra
muestra = len(df)
#Tabla con 4 frecuencias completa
tabla= pd.DataFrame({
    'fi': frecuencia,
    'hi (%)': (frecuencia/muestra * 100).round(2), # decimales(2)
    'FI': frecuencia.cumsum(),
    'HI (%)': (frecuencia/muestra * 100).cumsum().round(2)
})
tabla
```

### Tabla de frecuencia para variable **cuantitativa continua** de un DataFrame de nombre **df**

```python
#Agregar una columna al DataFrame df

k = cantidad de intervalos #Debe ingresar este valor
n = tamaño de la muestra #Debe ingresar este valor
d = cantidad de decimales de los intervalos #Debe ingresar este valor

df['Intervalos'] = pd.cut(df['Nombre de la variable'], bins = k, include_lowest= True, precision = d)
frecuencia = df.groupby('Intervalos', observed = True).size()

tabla_continua = pd.DataFrame({
    'fi': frecuencia,
    'hi': (frecuencia/n)*100,
    'Fi':frecuencia.cumsum(),
    'Hi': ((frecuencia/n) * 100).cumsum()

})
tabla_continua #Puede indicar redondear con m cantidad de decimales round(tabla_continua,m)
```

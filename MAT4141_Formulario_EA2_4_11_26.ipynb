### Modelo de regresión lineal usando los datos del DataFrame **df**

```python
#Defición de variables independiente y dependiente
v_indep = df['Nombre de la variable independiente']
v_depen = df['Nombre de la variable dependiente']

# Obtener la pendiente y el intercepto
modelo_lineal = np.polyfit(v_indep,v_depen,1)
pendiente, intercepto = modelo_lineal

# Mostrar los resultados
print(f'Pendiente: {pendiente:.2f}')
print(f'Intercepto: {intercepto:.2f}')

# Mostrar la ecuación de regresión
print(f'Modelo lineal: y = {pendiente:.2f}x {intercepto:+.2f}')
```

### Coeficiente de determinación $R^{2}$ de un modelo de regresión lineal

```python
# Definir las variables
v_indep = df['xxxx']
v_depen = df['xxxxxx']

# Calcular el modelo de regresión lineal
modelo_lineal = np.polyfit(v_indep, v_depen, 1)

# Calcular las ventas estimadas por el modelo
predicciones_xxx = np.polyval(modelo_lineal, v_indep)

# Calcular el coeficiente de determinación R²
R2 = r2_score(v_depen, predicciones_xxx)

# Mostrar el resultado en porcentaje
print(f'Coeficiente de determinación: {R2 * 100:.2f}%')
```

### Cálculo de probabilidades usando los datos del DataFrame **df** a PARTIR DE UNA TABLA DE FRECUENCIA

```python
tabla = pd.DataFrame({
    'Categoria': ['A', 'B', 'C', 'D'],  # Ingresamos las categorías del ejercicio
    'Frecuencia': [10, 20, 30, 40]   # Ingresamos las frecuencias de cada categoría
})

# PASO 2: Calculamos el total de observaciones
total = tabla['Frecuencia'].sum()  # Sumamos todas las frecuencias

# PASO 3: Calculamos las probabilidades
tabla['Probabilidad %'] = (tabla['Frecuencia'] / total) * 100  # Aplicamos la fórmula de probabilidad

# PASO 4: Mostramos los resultados
round(tabla, 1)  # Mostramos la tabla con las probabilidades redondeadas a un decimal
```

Código general: Identificación del resultado más probable y menos probable

```python
# PASO 1: Identificamos la categoría con mayor probabilidad
mas_probable = tabla.loc[tabla['Probabilidad %'].idxmax()]

# PASO 2: Identificamos la categoría con menor probabilidad
menos_probable = tabla.loc[tabla['Probabilidad %'].idxmin()]

# PASO 3: Mostramos los resultados para su interpretación

# Resultado más probable
print("RESULTADO MÁS PROBABLE")
print("Categoría:", mas_probable['Parentesco'])
print("Probabilidad:", round(mas_probable['Probabilidad %'], 3), "%")

# Resultado menos probable
print("RESULTADO MENOS PROBABLE")
print("Categoría:", menos_probable['Parentesco'])
print("Probabilidad:", round(menos_probable['Probabilidad %'], 3), "%")
```

```python
# PASO 1: Calculamos las probabilidades para dos variables
probabilidad = pd.crosstab(
    df['Variable_1'],       # Seleccionamos la primera variable (filas)
    df['Variable_2'],       # Seleccionamos la segunda variable (columnas)
    normalize='index',      # Definimos el tipo de probabilidad que queremos calcular
    margins=True            # Agregamos los totales correspondientes
) * 100                     # Convertimos las probabilidades en porcentajes

# PASO 2: Seleccionamos el tipo de probabilidad modificando normalize
# normalize='all'     → Probabilidad conjunta (respecto del total general)
# normalize='index'   → Probabilidad condicional por filas
# normalize='columns' → Probabilidad condicional por columnas

# PASO 3: Mostramos los resultados
round(probabilidad, 1)      # Mostramos las probabilidades redondeadas a un decimal
```
